# Projek Siscer

# **Load Data**

```python
import pandas as pd
df = pd.read_csv('Dataset_water.csv')
df
```

# **Data preparation**

### Data seperation x and y

```python
y = df['Dissolved Oxygen (mg/L)']
y
```

```python
x = df.drop(['Dissolved Oxygen (mg/L)','Sample ID'], axis = 1)
x
```

### Data splitting

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state=100)
```

```python
x_train
```

```python
x_test
```

# **Model Building**

## **Linear Regression**

**Training the model**

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(x_train, y_train)
```

**Apllying the model to make a prediction**

```python
y_lr_train_pred = lr.predict(x_train)
y_lr_test_pred = lr.predict(x_test)
```

```python
y_lr_train_pred
```

```python
y_lr_test_pred
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.scatter(y_train, lr.predict(x_train), color='blue', alpha=0.6)
plt.plot([y_train.min(), y_train.max()], [y_train.min(), y_train.max()], 'r--')
plt.xlabel('Nilai Aktual (DO)')
plt.ylabel('Prediksi Model')
plt.title('Training: Prediksi vs Aktual')
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Evaluate model performance**

```python
from sklearn.metrics import mean_squared_error, r2_score

lr_train_mse = mean_squared_error(y_train, y_lr_train_pred)
lr_train_r2 = r2_score(y_train, y_lr_train_pred)

lr_test_mse = mean_squared_error(y_test, y_lr_test_pred)
lr_test_r2 = r2_score(y_test, y_lr_test_pred)
```

```python
print('LR MSE (Train) :', lr_train_mse)
print('LR R2 (Train) :', lr_train_r2)
print('LR MSE (Test) :', lr_test_mse)
print('LR R2 (Test) :', lr_test_r2)
```

# **Coba input**

```python
x_input = [[7.25,23.1,4.5,342]]  # satu baris, empat kolom
y_pred = lr.predict(x_input)
print("Hasil prediksi:", y_pred)
```
